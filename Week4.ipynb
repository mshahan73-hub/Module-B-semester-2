```python
# Homework 3 Analysis: Logistics & Supply Chain Dataset (Optimized)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('logistics_and_supply_chain.csv')

# --- Step 1: Check dataset ---
print("Columns in the dataset:\n", df.columns)
print("\nFirst 5 rows:\n", df.head())
df.info()

# --- Step 2: Define numerical and categorical columns ---
num_cols = [
    'fuel_consumption_rate',
    'eta_variation_hours',
    'warehouse_inventory_level',
    'loading_unloading_time',
    'handling_equipment_availability',
    'shipping_costs',
    'supplier_reliability_score',
    'lead_time_days',
    'historical_demand',
    'iot_temperature',
    'customs_clearance_time',
    'driver_behavior_score',
    'fatigue_monitoring_score',
    'disruption_likelihood_score',
    'delay_probability',
    'delivery_time_deviation'
]

cat_cols = [
    'traffic_congestion_level',
    'order_fulfillment_status',
    'weather_condition_severity',
    'port_congestion_level',
    'cargo_condition_status',
    'route_risk_level',
    'risk_classification'
]

# --- Step 3: Create a sample for faster plotting ---
sample_df = df.sample(min(len(df), 2000), random_state=42)

# --- Step 4: Univariate Analysis (Histograms + KDE + Violin) ---
for col in num_cols:
    plt.figure(figsize=(8,4))
    sns.histplot(sample_df[col], bins=20, kde=True)
    plt.title(f'Histogram + KDE for {col}')
    plt.show()
    
    plt.figure(figsize=(8,4))
    sns.violinplot(x=sample_df[col])
    plt.title(f'Violin plot for {col} (sampled)')
    plt.show()

# --- Step 5: Summary statistics ---
for col in num_cols:
    print(f'Column: {col}')
    print(f'Mean: {df[col].mean():.2f}')
    print(f'Median: {df[col].median():.2f}')
    print(f'Mode: {df[col].mode()[0]}')
    print(f'Min: {df[col].min()} | Max: {df[col].max()}')
    print('-'*40)

# --- Step 6: Categorical Data: Bar Plots ---
for col in cat_cols:
    plt.figure(figsize=(8,4))
    sns.countplot(x=col, data=df)
    plt.title(f'Counts of {col}')
    plt.xticks(rotation=45)
    plt.show()

# --- Step 7: Grouped / Comparison Plot ---
# Replace slow swarm plot with boxplot for speed
plt.figure(figsize=(10,5))
sns.boxplot(x='traffic_congestion_level', y='delivery_time_deviation', data=sample_df)
plt.title('Delivery Time Deviation by Traffic Congestion Level (Boxplot)')
plt.show()

# --- Step 8: Effect of Bin Size on Histogram ---
col_example = 'delivery_time_deviation'

plt.figure(figsize=(8,4))
sns.histplot(sample_df[col_example], bins=5, kde=True)
plt.title(f'Histogram for {col_example} with 5 bins')
plt.show()

plt.figure(figsize=(8,4))
sns.histplot(sample_df[col_example], bins=50, kde=True)
plt.title(f'Histogram for {col_example} with 50 bins')
plt.show()

```

```python

# --- Step 9: Storytelling With Data Style Graph ---
# Stacked bar chart example
story_data = df.groupby(['traffic_congestion_level', 'risk_classification'])['delivery_time_deviation'].count().unstack()
story_data.plot(kind='bar', stacked=True, figsize=(10,6), color=['#4C72B0','#55A868','#C44E52','#8172B3'])
plt.title('Delivery Time Deviation by Traffic Congestion & Risk Classification', fontsize=14)
plt.xlabel('Traffic Congestion Level', fontsize=12)
plt.ylabel('Number of Deliveries', fontsize=12)
plt.xticks(rotation=0)
plt.legend(title='Risk Classification')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# --- Step 10: Optional Correlation Heatmap ---
plt.figure(figsize=(12,8))
sns.heatmap(df[num_cols].corr(), annot=True, fmt=".2f", cmap='coolwarm')
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```
